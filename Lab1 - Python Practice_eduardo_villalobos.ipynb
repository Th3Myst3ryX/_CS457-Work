#### Undergraduate student Instructions (CS457)

- Let's think if we have this function: f (t) defined by: f (t) = 2t / (1 + t ^ 2) ^ 2 with t> 0.
- Where t could be generated by np.linspace(0,10,100)
- Provide plot results

```python
# Your code:
def function_t(x):
        return 2*x/(1+x**2)
   
x_nums = np.linspace(0,10,100)
function_vals = function_t(x_nums)
```

```python
# Plot graph with x, y and label set the fontsize = 14
# Plot <Your Name> + title and set the fontsize = 14
# plot size as (8,5)
# plot with legend
# plot with grid
plt.figure(figsize=(8,5))
plt.plot(x_nums,function_vals, label="(function (t))", color = "blue", linewidth=2)

plt.xlabel("Input Values(x)",fontsize=14)
plt.ylabel("Function Output(y)",fontsize=14)
plt.title("Eduardo Villalobos + Graph of $y = \\frac{2x}{(1+x^2)^2}$", fontsize=14)
plt.grid(alpha=0.4)
plt.legend()

plt.show()
```
